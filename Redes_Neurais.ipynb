```python
import torch

lista = [[1,2,3],
        [4,5,6]]

tns = torch.Tensor(lista)
print(tns.dtype)
print(tns)

tns = torch.FloatTensor(lista)
print(tns.dtype)
print(tns)

tns = torch.DoubleTensor(lista)
print(tns.dtype)
print(tns)

tns = torch.LongTensor(lista)
print(tns.dtype)
print(tns)
```

```python
import numpy as np

arr = np.random.rand(3,4)
tns = torch.from_numpy(arr)

print(arr)
print(arr.dtype)

print(tns.dtype)
print(tns)
```

```python
tns1 = torch.ones(2,3)
tns0 = torch.zeros(4,5)
tnsr = torch.randn(3,3)

print(tns1)
print(tns0)
print(tnsr)
```

```python
print(tnsr)
tnsr[0,2] = -10

print("")
print(tnsr)
print("")
print(tnsr[:,2])

print(tnsr[0,2])
```

```python
#Operacoes com tensores
#Tudo feito ponto a ponto
#Ele multiplica de ponto a ponto e nao a multiplicacao de matrizes que esperamos
tns = tnsr[0:2, :]
print(tns)
print(tns1)
print("")
print(tns+tns1)
print("")
print(tns*tns1)
print("")
print(tns1/tns)
print("")
print(tns1//tns)
print("")
print(tns1%tns)
print("")

#Funcoes para fazer produto interno
print(torch.mm(tns1, tns.T))
```

```python
#Tamanho de um Tensor
tns = torch.randn(2,2,3)
print(tns)

print("")
print(tns.size())

print("")
#Achatar tensor
tns = tns.view(tns.size(0), -1) #O mais comum eh isso, manter a primeira dimensao e achatar o resto
#Poderia ser 12
#Ou -1 que tambem seria achatar

print(tns)
print(tns.size())

```

```python
tns = torch.randn(9, 12)
tns1 = tns[0:5, 0:4]
tns2 = tns[5:, 4:]

resultado = torch.mm(tns1, tns2)
print(resultado.size())
```

```python
tns1 = torch.randn(7,7,3)
tns2 = torch.randn(147, 1)

tns1 = tns1.view(-1, 1)
soma = tns1 + tns2
print(soma)
```

```python
#GPU Cast
#Serve para pegar dados e colocar na GPU
#Para um processamento mais rapido
#Verificar se a GPU esta disponivel

tns = torch.randn(10)

if torch.cuda.is_available():
  device = torch.device('cuda')
else:
  device = torch.device('cpu')
print(device)

tns = tns.to(device)
print(tns)
```
